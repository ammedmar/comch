# Cartan coboundaries: even and odd primes

This notebook uses `comch` to implement effective Cartan constructions in the Barratt-Eccles and surjection operads. We first review the even-prime construction of [Med20](#Med20), then pass to its odd-prime extension and low-degree cochain calculations.

## Introduction

For a space $X$, the Cartan formula relates the cup product and the Steenrod squares by

$$
Sq^k\big([\alpha][\beta]\big) = \sum_{r+s=k} Sq^r([\alpha])Sq^s([\beta]).
$$

Using cup-$i$ products, this relation is equivalent to the existence, for every $i \geq 0$ and pair of mod-$2$ cocycles $\alpha,\beta$, of a natural cochain $\zeta_i(\alpha \otimes \beta)$ satisfying

$$
\delta \zeta_i(\alpha \otimes \beta) =
(\alpha \smallsmile_0 \beta) \smallsmile_i (\alpha \smallsmile_0 \beta)
+ \sum_{i=j+k} (\alpha \smallsmile_j \alpha) \smallsmile_0 (\beta \smallsmile_k \beta).
$$

Such a cochain is called a **Cartan $i$-coboundary**.

This relation holds for all $E_\infty$-algebras and the purpose of this notebook is to explain how to construct elements in the arity four part of the Barratt-Eccles and surjection operad whose value on $(\alpha \otimes \alpha \otimes \beta \otimes \beta)$, where $\alpha$ and $\beta$ are mod $2$ cocycles, is a Cartan coboundary.

## The Barratt-Eccles Cartan homotopy

Let

$$
\widetilde{x}_i = (e,(12),e,\ldots,(12)^i) \in \mathcal E(2).
$$

The paper constructs degree-$1$ maps $H_1,H_2 : \mathcal E(2) \to \mathcal E(4)$ and proves that

$$
\zeta_i(\alpha \otimes \beta) =
(H_1+H_2)(\widetilde{x}_i)
(\alpha \otimes \alpha \otimes \beta \otimes \beta)
$$

is a Cartan $i$-coboundary. The method `BarrattEccles.cartan_homotopy(i)` returns the normalized Barratt-Eccles element $(H_1+H_2)(\widetilde{x}_i)$ over $\mathbb F_2$. We refer to this element as a **Cartan $i$-relator**.

```python
from comch import BarrattEccles

for i in range(3):
    print(f'(H1 + H2)(x_{i}) =')
    print(BarrattEccles.cartan_homotopy(i))
    print()
```

## Table reduction and surjection Cartan homotopies

The table reduction morphism $TR \colon \mathcal E \to \mathcal X$ sends these Barratt-Eccles elements to surjection elements, producing Cartan coboundaries for any algebra over the surjection operad.

<!-- 

Table reduction uses the Berger-Fresse convention. The simplicial action below relabels its output with the McClure-Smith convention; over $\mathbb F_2$ there is no sign issue. -->

```python
from comch import Surjection

for i in range(3):
    print(f'TR((H1 + H2)(x_{i})) =')
    print(Surjection.cartan_homotopy(i))
    print()
```

## Action on simplices

Simplicial cochains are algebras over both operads. The action is induced by the Alexander-Whitney coalgebra structure on simplicial chains.
For any two mod 2 cocycles, a Cartan $i$-coboundary is determined by the action of the Cartan $i$-relator on chains.

$$
\zeta_i(\alpha \otimes \beta)(c) =
(\alpha \otimes \alpha \otimes \beta \otimes \beta)z_i(c).
$$

We now describe $z_i$ on the first elements in the universal basis of simplicial chains.
We can restrict to homogeneous $\alpha$ and $\beta$, so we filter the output of $z_i$ accordingly.

## Examples

### Simplices

The smallest degree $n = i + 3$ for which the Cartan coboundary is not identically zero on homogeneous cocycles.

```python
from comch import Steenrod

for i in range(3):
    n = i + 3
    print(f'z_{i} on {list(range(n+1))} =')
    print(Steenrod.cartan_coboundaries(i, n, shape='simplex'))
    print()
```

## Odd-prime Cartan homotopies

The odd-prime construction implemented here follows Cantero-Morán and Medina-Mardones [CMM25](#CMM25).

For a cyclic arity $r\geq 2$, the effective Cartan relator is the degree-$1$ map

$$
H = K_1\iota + K_2\iota + \circ_{\mathcal E}(e\otimes K_3).
$$

Here $\iota\colon \mathcal W(r)\to\mathcal E(r)$ is the May--Steenrod map. The prism homotopy $K_1$ compares the interleaved and block-ordered products, $K_2$ is the signed Eilenberg--Zilber/Shih homotopy, and $K_3$ corrects the failure of $\iota$ to preserve the coproduct strictly. The third term vanishes when $r=2$ and first contributes in resolution degree $i=2$.

`BarrattEccles.cartan_homotopy(i, arity=r)` evaluates this full relator on the preferred generator $e_i$. It returns an element of $\mathcal E(2r)_{i+1}$ over $\mathbb Z/r$. For an odd prime, set `arity=p`.

```python
from comch import BarrattEccles, Surjection, Steenrod

p = 3
for i in range(2):
    print(f'H(e_{i}) in E(6):')
    print(BarrattEccles.cartan_homotopy(i, arity=p))
    print()
```

The displayed elements lie in $\mathcal E(6)_{i+1}$ over $\mathbb F_3$. In degrees $0$ and $1$ only $K_1+K_2$ occurs. The coproduct correction $K_3$ first appears on $e_2$.

### Table reduction and signs

The integral table-reduction map lands in the Berger--Fresse convention. At the even prime the two supported sign conventions coincide modulo $2$, but at odd primes they do not. Consequently, odd Cartan homotopies currently require `convention='Berger-Fresse'`; relabeling the same coefficients as McClure--Smith would be mathematically incorrect.

```python
for i in range(2):
    print(f'TR(H(e_{i})) in X(6):')
    print(Surjection.cartan_homotopy(
        i, arity=p, convention='Berger-Fresse'))
    print()
```

### Universal simplicial representatives

The high-level call `Steenrod.cartan_coboundaries(i, n, arity=p)` acts with the odd Cartan homotopy on the standard $n$-simplex and retains the bihomogeneous summands. Its output has $2p$ tensor factors: the first $p$ have a common degree and the last $p$ have a common degree.

```python
odd_z0 = Steenrod.cartan_coboundaries(0, 2, arity=3)
print(odd_z0)
```

### The Cartan-relator identity

The defining property of the construction is not a term count but the chain-homotopy equation

$$
\partial H(e_i)+H(\partial e_i)=G(e_i)-\tau F(e_i).
$$

Here $\tau F$ first multiplies and then applies the May--Steenrod operation, whereas $G$ applies the cyclic diagonal and then multiplies. The following low-degree computation forms these maps independently and compares the two sides in the Barratt--Eccles operad. The mod-$2$ and mod-$3$ columns therefore express the same homotopical statement, with the appropriate cyclic resolutions.

```python
from comch.barratt_eccles import BarrattEcclesElement
from comch.symmetric import SymmetricGroupElement, SymmetricRingElement

def right_action(element, permutation, exponent=1):
    return element.create({
        tuple(value * (permutation ** exponent) for value in simplex): coefficient
        for simplex, coefficient in element.items()
    })

def odd_cartan_endpoints(degree, p=3):
    rotation = SymmetricGroupElement(tuple(range(2, p + 1)) + (1,))
    product = BarrattEccles.may_steenrod_map(2, 0, torsion=p)
    source = BarrattEccles.may_steenrod_map(p, degree, torsion=p)
    tau = SymmetricGroupElement(
        tuple(range(1, 2 * p, 2)) + tuple(range(2, 2 * p + 1, 2)))

    reordered = BarrattEcclesElement(torsion=p)
    for simplex, coefficient in source.items():
        image = []
        for permutation in simplex:
            vertex = BarrattEcclesElement({(permutation,): 1}, torsion=p)
            for position in range(1, 2 * p, 2):
                vertex = vertex.compose(product, position)
            image.append(next(iter(vertex))[0] * tau)
        reordered += reordered.create({tuple(image): coefficient})

    half_degree = degree // 2
    diagonal_terms = []
    if degree % 2 == 0:
        diagonal_terms.extend(
            (2 * j, 2 * (half_degree - j), 0, 0)
            for j in range(half_degree + 1))
        diagonal_terms.extend(
            (2 * j + 1, 2 * (half_degree - 1 - j) + 1, s, t)
            for j in range(half_degree)
            for s in range(p) for t in range(s + 1, p))
    else:
        for j in range(half_degree + 1):
            diagonal_terms.extend((
                (2 * j, 2 * (half_degree - j) + 1, 0, 0),
                (2 * j + 1, 2 * (half_degree - j), 0, 1)))

    cyclic = BarrattEcclesElement(torsion=p)
    for left_degree, right_degree, s, t in diagonal_terms:
        left = right_action(BarrattEccles.may_steenrod_map(
            p, left_degree, torsion=p), rotation, s)
        right = right_action(BarrattEccles.may_steenrod_map(
            p, right_degree, torsion=p), rotation, t)
        cyclic += product.compose(left, 1).compose(right, p + 1)
    return reordered, cyclic

def odd_relator_sides(degree, p=3):
    reordered, cyclic = odd_cartan_endpoints(degree, p)
    lhs = BarrattEccles.cartan_homotopy(degree, arity=p).boundary()
    if degree:
        rotation = SymmetricGroupElement(tuple(range(2, p + 1)) + (1,))
        block_rotation = SymmetricGroupElement(
            tuple(rotation) + tuple(p + value for value in rotation))
        previous = BarrattEccles.cartan_homotopy(degree - 1, arity=p)
        if degree % 2:
            lhs += right_action(previous, block_rotation) - previous
        else:
            for exponent in range(p):
                lhs += right_action(previous, block_rotation, exponent)
    return lhs, cyclic - reordered

def even_relator_sides(degree):
    x_0 = BarrattEccles.may_steenrod_map(2, 0, torsion=2)
    x_i = BarrattEccles.may_steenrod_map(2, degree, torsion=2)
    tau = SymmetricRingElement({(1, 3, 2, 4): 1}, torsion=2)
    reordered = tau * x_i.compose(x_0, 1).compose(x_0, 3)
    diagonal = BarrattEcclesElement(torsion=2)
    for (left, right), coefficient in x_i.diagonal().items():
        left = BarrattEcclesElement({left: 1}, torsion=2)
        right = BarrattEcclesElement({right: 1}, torsion=2)
        diagonal += coefficient * x_0.compose(left, 1).compose(right, 3)
    lhs = BarrattEccles.cartan_homotopy(degree).boundary()
    if degree:
        resolution_boundary = SymmetricRingElement({
            (2, 1, 4, 3): 1, (1, 2, 3, 4): 1}, torsion=2)
        lhs += resolution_boundary * BarrattEccles.cartan_homotopy(degree - 1)
    return lhs, reordered + diagonal

print('degree | mod 2 | mod 3')
print('-------+-------+------')
for degree in range(4):
    even_lhs, even_rhs = even_relator_sides(degree)
    odd_lhs, odd_rhs = odd_relator_sides(degree)
    print(f'{degree:6} | {even_lhs == even_rhs!s:5} | {odd_lhs == odd_rhs!s:5}')
```

### Why the odd correction $K_3$ is necessary

Let $\Delta_{\mathcal W}$ be the cyclic diagonal and $\Delta_{AW}$ the Alexander--Whitney diagonal. At the even prime the May--Steenrod map is compatible with these diagonals in the required construction, and the correction vanishes. At an odd prime there is a genuine defect

$$
G(e_2)-G_{AW}(e_2),
$$

where $G_{AW}$ uses $\Delta_{AW}\iota(e_2)$. The first nonzero $K_3$ occurs precisely in this degree and satisfies

$$
\partial\big(\circ(e\otimes K_3(e_2))\big)
=G(e_2)-G_{AW}(e_2).
$$

No $K_3(\partial e_2)$ term appears because $K_3(e_1)=0$. Thus this calculation isolates the new odd-prime phenomenon rather than checking only the total homotopy.

```python
def alexander_whitney_product(degree, p):
    product = BarrattEccles.may_steenrod_map(2, 0, torsion=p)
    source = BarrattEccles.may_steenrod_map(p, degree, torsion=p)
    answer = BarrattEcclesElement(torsion=p)
    for (left, right), coefficient in source.diagonal().items():
        left = BarrattEcclesElement({left: 1}, torsion=p)
        right = BarrattEcclesElement({right: 1}, torsion=p)
        answer += coefficient * product.compose(left, 1).compose(right, p + 1)
    return answer

even_k3 = BarrattEccles.cartan_coproduct_correction(2, arity=2)
odd_k3 = BarrattEccles.cartan_coproduct_correction(2, arity=3)
_, cyclic_product = odd_cartan_endpoints(2, p=3)
coproduct_defect = cyclic_product - alexander_whitney_product(2, p=3)

print('K3(e_2) at p=2:', even_k3)
print('number of terms in the composed K3(e_2) at p=3:', len(odd_k3))
print('number of terms in its boundary:', len(odd_k3.boundary()))
print('boundary(K3 correction) equals the coproduct defect:',
      odd_k3.boundary() == coproduct_defect)
```
