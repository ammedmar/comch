# Cartan coboundaries: the even prime

This notebook follows the paper [*An effective proof of the Cartan formula: the even prime*](https://doi.org/10.1016/j.jpaa.2020.106444) and uses `comch` to implement its construction in the Barratt-Eccles and surjection operads.

## Introduction

For a space $X$, the Cartan formula relates the cup product and the Steenrod squares by

$$
Sq^k\big([\alpha][\beta]\big) = \sum_{r+s=k} Sq^r([\alpha])Sq^s([\beta]).
$$

Using cup-$i$ products, this relation is equivalent to the existence, for every $i \geq 0$ and pair of mod-$2$ cocycles $\alpha,\beta$, of a natural cochain $\zeta_i(\alpha \otimes \beta)$ satisfying

$$
\delta \zeta_i(\alpha \otimes \beta) =
(\alpha \smallsmile_0 \beta) \smallsmile_i (\alpha \smallsmile_0 \beta)
+ \sum_{i=j+k} (\alpha \smallsmile_j \alpha) \smallsmile_0 (\beta \smallsmile_k \beta).
$$

Such a cochain is called a **Cartan $i$-coboundary**.

This relation holds for all $E_\infty$-algebras and the purpose of this notebook is to explain how to construct elements in the arity four part of the Barratt-Eccles and surjection operad whose value on $(\alpha \otimes \alpha \otimes \beta \otimes \beta)$, where $\alpha$ and $\beta$ are mod $2$ cocycles, is a Cartan coboundary.

## The Barratt-Eccles Cartan homotopy

Let

$$
\widetilde{x}_i = (e,(12),e,\ldots,(12)^i) \in \mathcal E(2).
$$

The paper constructs degree-$1$ maps $H_1,H_2 : \mathcal E(2) \to \mathcal E(4)$ and proves that

$$
\zeta_i(\alpha \otimes \beta) =
(H_1+H_2)(\widetilde{x}_i)
(\alpha \otimes \alpha \otimes \beta \otimes \beta)
$$

is a Cartan $i$-coboundary. The method `BarrattEccles.cartan_homotopy(i)` returns the normalized Barratt-Eccles element $(H_1+H_2)(\widetilde{x}_i)$ over $\mathbb F_2$. We refer to this element as a **Cartan $i$-relator**.

```python
from comch import BarrattEccles

for i in range(3):
    print(f'(H1 + H2)(x_{i}) =')
    print(BarrattEccles.cartan_homotopy(i))
    print()
```

## Table reduction and surjection Cartan homotopies

The table reduction morphism $TR \colon \mathcal E \to \mathcal X$ sends these Barratt-Eccles elements to surjection elements, producing Cartan coboundaries for any algebra over the surjection operad.

<!-- 

Table reduction uses the Berger-Fresse convention. The simplicial action below relabels its output with the McClure-Smith convention; over $\mathbb F_2$ there is no sign issue. -->

```python
from comch import Surjection

for i in range(3):
    print(f'TR((H1 + H2)(x_{i})) =')
    print(Surjection.cartan_homotopy(i))
    print()
```

## Action on simplices and cubes

Simplicial and cubical cochains are algebras over both operads.
The action is induced by the Alexander-Whitney coalgebra structure on simplicial chains and by the Serre coalgebra structure on cubical chains.
For any two mod 2 cocycles, a Cartan $i$-coboundary is determined by the action of the Cartan $i$-relator on chains.

$$
\zeta_i(\alpha \otimes \beta)(c) =
(\alpha \otimes \alpha \otimes \beta \otimes \beta)z_i(c).
$$

We now describe $z_i$ on the first elements in the universal bases of simplicial and cubical chains.
We can restrict to homogeneous $\alpha$ and $\beta$, so we filter the output of $z_i$ accordingly.

## Examples

### Simplices

The smallest degree $n = i + 3$ for which the Cartan coboundary is not identically zero on homogeneous cocycles.

```python
from comch import Steenrod

for i in range(3):
    n = i + 3
    print(f'z_{i} on {list(range(n+1))} =')
    print(Steenrod.cartan_coboundaries(i, n, shape='simplex'))
    print()
```
