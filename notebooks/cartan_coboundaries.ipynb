# An effective construction of Cartan coboundaries

This notebook follows the paper [*An effective proof of the Cartan formula: the even prime*](https://doi.org/10.1016/j.jpaa.2020.106444) and uses `comch` to implement its construction in the Barratt-Eccles operad.

## Introduction

For a space $X$, the Cartan formula relates the cup product and the Steenrod squares by

$$
Sq^k\big([\alpha][\beta]\big) = \sum_{r+s=k} Sq^r([\alpha])Sq^s([\beta]).
$$

Using cup-$i$ products, this relation is equivalent to the existence, for every $i \geq 0$ and pair of mod-$2$ cocycles $\alpha,\beta$, of a natural cochain $\zeta_i(\alpha \otimes \beta)$ satisfying

$$
\delta \zeta_i(\alpha \otimes \beta) =
(\alpha \smallsmile_0 \beta) \smallsmile_i (\alpha \smallsmile_0 \beta)
+ \sum_{i=j+k} (\alpha \smallsmile_j \alpha) \smallsmile_0 (\beta \smallsmile_k \beta).
$$

Such a cochain is called a **Cartan $i$-coboundary**.

## The Barratt-Eccles Cartan homotopy

Let

$$
\widetilde{x}_i = (e,(12),e,\ldots,(12)^i) \in \mathcal E(2).
$$

The paper constructs degree-$1$ maps $H_1,H_2 : \mathcal E(2) \to \mathcal E(4)$ and proves that

$$
\zeta_i(\alpha \otimes \beta) =
(H_1+H_2)(\widetilde{x}_i)
(\alpha \otimes \alpha \otimes \beta \otimes \beta)
$$

is a Cartan $i$-coboundary. The method `BarrattEccles.cartan_homotopy(i)` returns the normalized Barratt-Eccles element $(H_1+H_2)(\widetilde{x}_i)$ over $\mathbb F_2$.

```python
from comch import BarrattEccles

for i in range(3):
    print(f'(H1 + H2)(x_{i}) =')
    print(BarrattEccles.cartan_homotopy(i))
    print()
```

## Table reduction and simplicial chains

The table reduction morphism $TR : \mathcal E \to \mathcal X$ sends these Barratt-Eccles elements to surjection elements. For a simplicial set $X$, their action determines maps

$$
z_i : N_n(X;\mathbb F_2) \longrightarrow N_*(X;\mathbb F_2)^{\otimes 4}
$$

with

$$
\zeta_i(\alpha \otimes \beta)(c) =
(\alpha \otimes \alpha \otimes \beta \otimes \beta)z_i(c).
$$

Table reduction uses the Berger-Fresse convention. The simplicial action below relabels its output with the McClure-Smith convention; over $\mathbb F_2$ there is no sign issue.

```python
from comch import SurjectionElement

for i in range(3):
    print(f'TR((H1 + H2)(x_{i})) =')
    print(BarrattEccles.cartan_homotopy(i).table_reduction())
    print()
```

```python
def cartan_coboundary_chain(i, n):
    """Return the homogeneous part of z_i on the top n-simplex."""
    surjection = BarrattEccles.cartan_homotopy(i).table_reduction()
    surjection = SurjectionElement(
        surjection, torsion=2, convention='McClure-Smith')
    simplex = Simplicial.standard_element(n, torsion=2)
    answer = surjection(simplex)
    homogeneous = {
        key: value for key, value in answer.items()
        if key[0].dimension == key[1].dimension
        and key[2].dimension == key[3].dimension
    }
    return answer.create(homogeneous)
```

## Examples

Let $\mathrm{id}_n = (0,\ldots,n)$ be the top-dimensional basis element of the standard $n$-simplex. The following reproduces the explicit examples in the paper for $i=0,1,2$, using the smallest dimension $n=i+3$ for which the Cartan coboundary is not identically zero on homogeneous cocycles.

```python
from comch import Simplicial

or i in range(3):
    n = i + 3
    print(f'z_{i}(id_{n}) =')
    print(cartan_coboundary_chain(i, n))
    print()
```
